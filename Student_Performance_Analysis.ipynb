```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("StudentsPerformance.csv", encoding='latin1')
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df[['math score','reading score','writing score']].mean()
```

```python
df.groupby('gender')[['math score','reading score','writing score']].mean()
```

```python
df.groupby('parental level of education')['math score'].mean()
```

```python
df.groupby('gender')['math score'].mean().plot(kind='bar')
plt.title("Math Score by Gender")
plt.show()
```

## 🤖 AI Insights

### 📌 Key Insights from Data Analysis

1. Male students have higher average math scores compared to female students.
2. Female students perform better in reading and writing subjects.
3. Students with higher parental education tend to score better.
4. Students with standard lunch show better academic performance.
5. Most students score in the range of 60–80, indicating average performance level.
